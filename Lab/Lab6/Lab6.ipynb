## 🧮 Lab 6.1: Simple Calculator Functions

### Objective:
Define and utilize functions to perform basic arithmetic operations, practicing parameter passing, return values, docstrings, and control flow integration.

### Requirements:
1. Define four basic arithmetic functions:
   - `add(a, b)`: Returns sum of $a + b$
   - `subtract(a, b)`: Returns difference of $a - b$
   - `multiply(a, b)`: Returns product of $a \times b$
   - `divide(a, b)`: Returns quotient of $a / b$, safely handling division by zero ($b = 0$)
2. Add a comprehensive **docstring** to each function explaining its purpose, parameters, and return value according to PEP 257.
3. **Challenge Task:** Implement `power(base, exponent=2)` using a default parameter so that omitting the exponent squares the base.
4. Implement an interactive **menu-driven program** (using a `while` loop) allowing users to select operations, enter numbers, and view results.


````python
# ==========================================
# Lab 6.1: Simple Calculator Function Definitions
# ==========================================

def add(a: float, b: float) -> float:
    """
    Calculate the sum of two numbers.

    Parameters:
    a (float or int): The first number (addend).
    b (float or int): The second number (addend).

    Returns:
    float: The sum of a and b.
    """
    return a + b


def subtract(a: float, b: float) -> float:
    """
    Calculate the difference between two numbers (a - b).

    Parameters:
    a (float or int): The minuend.
    b (float or int): The subtrahend to subtract from a.

    Returns:
    float: The difference (a - b).
    """
    return a - b


def multiply(a: float, b: float) -> float:
    """
    Calculate the product of two numbers.

    Parameters:
    a (float or int): The first factor (multiplicand).
    b (float or int): The second factor (multiplier).

    Returns:
    float: The product of a and b.
    """
    return a * b


def divide(a: float, b: float):
    """
    Calculate the quotient of two numbers (a / b).
    Safely handles division by zero by returning an error message string.

    Parameters:
    a (float or int): The dividend.
    b (float or int): The divisor.

    Returns:
    float or str: The quotient of a and b, or an error message if b == 0.
    """
    if b == 0:
        return "Error: Division by zero"
    return a / b


# Challenge: power function with default exponent=2
def power(base: float, exponent: float = 2) -> float:
    """
    Calculate the base raised to the specified exponent.
    If exponent is not provided, defaults to 2 (squaring the base).

    Parameters:
    base (float or int): The base value.
    exponent (float or int, optional): The power exponent. Defaults to 2.

    Returns:
    float: The result of base ** exponent.
    """
    return base ** exponent


print("All calculator functions and docstrings defined successfully.")
````

### 📱 Menu-Driven Calculator Application
Here we implement a menu-driven calculator program that runs inside a `while` loop, as studied in Week 3.
- The user can repeatedly choose operations until they select **6. Exit**.
- For Option 5 (Power), if the user presses **Enter** without typing an exponent, Python automatically utilizes the default parameter `exponent=2`!
- We also provide an automated demonstration simulating interactive user sessions.


````python
# ==========================================
# Menu-Driven Calculator Program
# ==========================================

def run_calculator(interactive: bool = False, simulated_inputs: list = None):
    """
    Runs the menu-driven calculator loop.
    Supports both live user interaction (interactive=True)
    and scripted simulation for notebook automated testing.
    """
    sim_index = 0
    def get_input(prompt_msg: str) -> str:
        nonlocal sim_index
        if interactive:
            return input(prompt_msg)
        else:
            if simulated_inputs and sim_index < len(simulated_inputs):
                val = simulated_inputs[sim_index]
                sim_index += 1
                print(f"{prompt_msg}{val}")
                return str(val)
            return "6"

    print("========================================")
    print("      Simple Calculator Program         ")
    print("========================================")

    while True:
        print("\nPlease select an operation:")
        print("  1. Addition (+)")
        print("  2. Subtraction (-)")
        print("  3. Multiplication (*)")
        print("  4. Division (/)")
        print("  5. Power (^) [Challenge - Default: 2]")
        print("  6. Exit Program")

        choice = get_input("Enter choice (1-6): ").strip()

        if choice == "6":
            print("\nThank you for using Simple Calculator. Goodbye!")
            break

        if choice not in ["1", "2", "3", "4", "5"]:
            print("[!] Invalid selection! Please enter a number between 1 and 6.")
            continue

        try:
            if choice == "5":
                base_str = get_input("Enter base number: ").strip()
                base_val = float(base_str)
                exp_str = get_input("Enter exponent (press Enter for default 2): ").strip()

                if exp_str == "":
                    # Uses default exponent=2
                    result = power(base_val)
                    print(f"\n[Result] power({base_val}) [squared] = {result}")
                else:
                    exp_val = float(exp_str)
                    result = power(base_val, exp_val)
                    print(f"\n[Result] {base_val} ^ {exp_val} = {result}")

            else:
                num1_str = get_input("Enter first number: ").strip()
                num2_str = get_input("Enter second number: ").strip()
                num1 = float(num1_str)
                num2 = float(num2_str)

                if choice == "1":
                    res = add(num1, num2)
                    print(f"\n[Result] {num1} + {num2} = {res}")
                elif choice == "2":
                    res = subtract(num1, num2)
                    print(f"\n[Result] {num1} - {num2} = {res}")
                elif choice == "3":
                    res = multiply(num1, num2)
                    print(f"\n[Result] {num1} * {num2} = {res}")
                elif choice == "4":
                    res = divide(num1, num2)
                    print(f"\n[Result] {num1} / {num2} = {res}")

        except ValueError:
            print("[Error] Invalid numeric input! Please enter valid numbers.")


# Run simulation demonstrating choices 1, 4 (div by zero), 5 (default exp), 5 (custom exp), and 6 (exit)
sample_session = [
    "1", "45", "15",       # Addition: 45 + 15 = 60
    "4", "10", "0",        # Division: 10 / 0 = Error
    "5", "8", "",          # Power: 8^2 (default) = 64
    "5", "2", "8",         # Power: 2^8 = 256
    "6"                    # Exit
]

run_calculator(interactive=False, simulated_inputs=sample_session)
````

### 🔍 Selected Lab Assignment: Week 2 (`Lab/Lab2/Lab2_1.py`) & Week 3 Loops
In **Week 2** ([`Lab/Lab2/Lab2_1.py`](../Lab2/Lab2_1.py)), we wrote a program to determine if an integer is positive/negative/zero and even/odd:

```python
# ========================================================
# Original Code from Lab/Lab2/Lab2_1.py
# ========================================================
# เขียนโปรแกรมที่รับตัวเลขจำนวนเต็มจากผู้ใช้ 1 ค่า แล้วแสดงผลว่าตัวเลขนั้นเป็นบวก ลบ หรือศูนย์ และเป็นเลขคู่หรือคี่ 
# โดยรวมผลลัพธ์ทั้งสองในประโยคเดียว เช่น "The number is negative and odd."

def predict_number(number):
    try:
        number = int(number)
    except ValueError:
        return "invalid input"
    calculate = ("positive" if number > 0 else "negative" if number < 0 else "zero") + " and " + ("even" if number % 2 == 0 else "odd")
    return "The number is " + calculate

number = input("Enter an integer: ")
print("====================================")
result = predict_number(number)
print(result)
```

#### ❌ Issues Identified in the Original Implementation:
1. **Coupled Business Logic:** The single line `calculate = ("positive" if number > 0 else "negative" if number < 0 else "zero") + " and " + ("even" if number % 2 == 0 else "odd")` compresses two independent domain rules (Sign check and Parity check) into one string expression.
2. **Poor Reusability:** If another module only needs to check parity (`is_even_odd`) or only sign (`is_positive_negative_zero`), it cannot do so without re-writing the conditional or parsing the string output.
3. **Limited to Single User Input:** The script only handles a single number at a time via `input()`. In **Week 3** ([`Lab/Lab3/Lab3.ipynb`](../Lab3/Lab3.ipynb)), we learned about loops (`for` and `while`). Without modular functions, processing a list or stream of numbers would require copy-pasting the ternary logic into loops.
4. **Untestable Intermediate Steps:** Because the results are immediately concatenated into a sentence, unit testing individual conditions is difficult.

---

#### ✅ Refactored Architecture (Applying Modular Design):
Following modular design principles and the assignment suggestions:
1. `is_positive_negative_zero(num)`: Dedicated sign classification returning `"positive"`, `"negative"`, or `"zero"`.
2. `is_even_odd(num)`: Dedicated parity classification returning `"even"` or `"odd"`.
3. `check_divisibility(num, d1, d2)`: Modular helper for divisibility checks.
4. `predict_number(number)`: Refactored version of the Week 2 function that cleanly **composes** `is_positive_negative_zero` and `is_even_odd`.
5. `classify_number(num)`: Returns a structured dictionary for data analysis pipelines.
6. `analyze_number_dataset(dataset)`: Incorporates **Week 3 Loops** (`for` loop) to process sequences of numbers with summary statistics.


````python
# ==========================================
# Activity 6.1: Refactored Modular Code
# (Refactoring Lab/Lab2/Lab2_1.py with Week 3 Loops)
# ==========================================

from typing import Dict, List, Any, Union

def is_positive_negative_zero(num: float) -> str:
    """
    Determines whether a number is positive, negative, or zero.
    
    Parameters:
    num (float or int): Number to evaluate.
    
    Returns:
    str: 'positive', 'negative', or 'zero'.
    """
    if num > 0:
        return "positive"
    elif num < 0:
        return "negative"
    return "zero"


def is_even_odd(num: int) -> str:
    """
    Determines whether an integer is even or odd.
    
    Parameters:
    num (int): Integer to evaluate.
    
    Returns:
    str: 'even' or 'odd'.
    """
    return "even" if int(num) % 2 == 0 else "odd"


def check_divisibility(num: int, d1: int = 3, d2: int = 5) -> str:
    """
    Checks divisibility of an integer by two divisors (default: 3 and 5).
    """
    int_n = int(num)
    by_d1 = (int_n % d1 == 0)
    by_d2 = (int_n % d2 == 0)
    
    if by_d1 and by_d2:
        return f"Divisible by {d1} and {d2}"
    elif by_d1:
        return f"Divisible by {d1}"
    elif by_d2:
        return f"Divisible by {d2}"
    return f"Not divisible by {d1} or {d2}"


# Refactored version of the original predict_number from Lab2_1.py
def predict_number(number: Any) -> str:
    """
    Refactored version of predict_number from Week 2 (Lab2_1.py).
    Composes modular helper functions instead of dense ternary chains.
    """
    try:
        val = int(number)
    except (ValueError, TypeError):
        return "invalid input"
    
    sign = is_positive_negative_zero(val)
    parity = is_even_odd(val)
    return f"The number is {sign} and {parity}."


def classify_number(num: int) -> Dict[str, Any]:
    """
    Composes individual classifications into a structured record.
    """
    return {
        "number": num,
        "sign": is_positive_negative_zero(num),
        "parity": is_even_odd(num),
        "divisibility": check_divisibility(num)
    }


def analyze_number_dataset(dataset: List[int]) -> None:
    """
    Demonstrates integration of modular functions with Week 3 loops (for loop).
    Processes a list of numbers and calculates summary metrics.
    """
    print(f"{'Number':>8} | {'Sign':<10} | {'Parity':<6} | {'Divisibility'}")
    print("-" * 55)
    
    pos_count = 0
    even_count = 0
    
    for n in dataset:
        record = classify_number(n)
        print(f"{record['number']:>8} | {record['sign']:<10} | {record['parity']:<6} | {record['divisibility']}")
        if record['sign'] == "positive":
            pos_count += 1
        if record['parity'] == "even":
            even_count += 1
            
    print("-" * 55)
    print(f"Summary Statistics: Total = {len(dataset)}, Positive = {pos_count}, Even = {even_count}")


# 1. Test Refactored predict_number (matching Week 2 specification)
print("=== 1. Testing Refactored predict_number (Week 2 Refactor) ===")
test_inputs = ["-7", "12", "0", "abc"]
for inp in test_inputs:
    print(f"predict_number('{inp}') -> {predict_number(inp)}")

# 2. Test Dataset Analysis with Week 3 Loops
print("\n=== 2. Testing Dataset Analysis with Week 3 Loops ===")
sample_dataset = [15, -7, 0, 30, -14, 22, -45, 8]
analyze_number_dataset(sample_dataset)
````

### 💡 Key Benefits Demonstrated by Refactoring:
1. **Single Responsibility Principle (SRP):** Each function does exactly one logical job (`is_positive_negative_zero`, `is_even_odd`, `check_divisibility`).
2. **Reusability & Extensibility:** If another part of an AI project requires checking if an input is even/odd or positive, `is_even_odd(n)` can be imported directly without dragging along printing logic.
3. **Testability:** Each function returns pure data (`str`, `dict`, `bool`), meaning we can write automated unit tests with simple assertion checks like `assert is_even_odd(4) == "Even"`.
4. **Readability & Clean Code:** The main processing loop `analyze_number_dataset` is concise and easy to understand at a glance.

## 💬 Activity 6.2: Discussion: Benefits of Modular Programming & Standard Library Exploration

### 1. Discussion: Benefits of Modular Programming
Modular programming is a software design technique that separates functionality into independent, interchangeable modules, each containing everything necessary to execute a specific aspect of the desired functionality.

#### 🌟 Key Scenarios & Benefits:

| Benefit | Scenario in Practice | Real-world Impact |
| :--- | :--- | :--- |
| **1. Code Reusability (DRY)** | A company has 10 data processing pipelines that all need to validate email addresses and clean timestamps. | By putting validators into a module `validation_utils.py`, developers write it once and import it across all 10 pipelines. Changes only need to be made in one place. |
| **2. Collaborative Teamwork** | A team of 5 software engineers is building an AI web application. | Engineer A builds `data_loader.py`, Engineer B builds `model.py`, Engineer C develops `api_routes.py`. Because interfaces (functions) are agreed upon, everyone works concurrently without merge conflicts. |
| **3. Isolated Debugging & Maintenance** | An error occurs when dividing amounts during checkout. | Instead of reading a 2,000-line monolithic script, the developer navigates directly to `calculate_discount()` in `finance.py`, isolates the bug, fixes it, and knows the rest of the application remains unaffected. |
| **4. Rigorous Unit Testing** | Continuous Integration / Continuous Deployment (CI/CD). | Automated testing frameworks (`pytest`, `unittest`) can test discrete functions (`is_prime(13) == True`) automatically on every code commit. Monolithic code with print statements cannot be automatically tested. |
| **5. Abstraction & Reduced Cognitive Load** | Using complex algorithms (e.g. Dijkstra's shortest path or image convolution). | The programmer calling `find_shortest_path(graph, start, end)` only needs to know the function inputs and return value, not the internal 200 lines of graph theory code. |

### 2. Python Standard Library Exploration
Python is famously described as having **"Batteries Included"**, meaning its Standard Library comes pre-packaged with rich, production-grade modules that solve common programming challenges without requiring third-party `pip` installations.

🔗 **Official Documentation:** [Python 3 Standard Library Documentation](https://docs.python.org/3/library/)

#### Three Explored Modules:
1. **`datetime` Module:**
   - **Purpose:** Supplies classes for manipulating dates and times (arithmetic, formatting, ISO standards).
   - **Use Case:** Logging timestamps in AI training loops, scheduling cron jobs, calculating user age or transaction expiration times.
2. **`collections` Module:**
   - **Purpose:** Provides specialized container datatypes beyond standard `dict`, `list`, `set`, and `tuple` (e.g., `Counter`, `defaultdict`, `deque`, `namedtuple`).
   - **Use Case:** Counting word frequencies in Natural Language Processing (NLP), tracking graph adjacency lists with `defaultdict(list)`, and managing queues with `deque`.
3. **`pathlib` Module (Modern replacement for `os.path`):**
   - **Purpose:** Offers object-oriented, cross-platform filesystem path representations.
   - **Use Case:** Managing dataset folders, safe cross-platform path concatenation (Windows `\` vs Linux `/`), checking file existence, and reading/writing file bytes without boilerplate.


````python
# ==========================================
# Activity 6.2: Practical Demonstration of Standard Library Modules
# ==========================================

# 1. datetime module demonstration
from datetime import datetime, timedelta

print("=== 1. Standard Library: datetime ===")
current_time = datetime.now()
print(f"Current Date & Time: {current_time.strftime('%Y-%m-%d %H:%M:%S')}")
future_deadline = current_time + timedelta(days=7, hours=3)
print(f"Project Deadline (+7 days, 3 hours): {future_deadline.strftime('%A, %d %B %Y at %I:%M %p')}")
elapsed = future_deadline - current_time
print(f"Time Delta: {elapsed.days} days and {elapsed.seconds // 3600} hours remaining.")

# 2. collections module demonstration (Counter & defaultdict)
from collections import Counter, defaultdict

print("\n=== 2. Standard Library: collections ===")
# Frequency counting in NLP / AI tokens
tokens = ["python", "ai", "machine", "learning", "python", "script", "ai", "python", "model"]
token_counts = Counter(tokens)
print(f"Word Frequencies: {dict(token_counts)}")
print(f"Top 2 Most Common Words: {token_counts.most_common(2)}")

# Grouping items with defaultdict
student_grades = [("Alice", "A"), ("Bob", "B"), ("Charlie", "A"), ("Diana", "C"), ("Evan", "B")]
grouped_by_grade = defaultdict(list)
for student, grade in student_grades:
    grouped_by_grade[grade].append(student)
print(f"Students grouped by grade: {dict(grouped_by_grade)}")

# 3. pathlib module demonstration
from pathlib import Path

print("\n=== 3. Standard Library: pathlib ===")
# Locate my_utils.py regardless of whether cwd is Lab/Lab6 or project root
sample_file = Path("my_utils.py") if Path("my_utils.py").exists() else Path("Lab/Lab6/my_utils.py")
target_dir = sample_file.parent.resolve() if sample_file.exists() else Path(".").resolve()
print(f"Directory: {target_dir.name}")
lab_files = list(target_dir.glob("*.py"))
print(f"Python (.py) files in directory: {[f.name for f in lab_files]}")
print(f"Does '{sample_file.name}' exist? {sample_file.exists()} | File size: {sample_file.stat().st_size} bytes")
````

ในการทดลองปฏิบัติการนี้ เราได้เรียนรู้และบรรลุความเข้าใจในประเด็นสำคัญต่าง ๆ อย่างครอบคลุม ดังนี้:

1. **การออกแบบฟังก์ชัน (Function Design):** การรวบรวมตรรกะและกระบวนการทางคณิตศาสตร์เข้าไว้ในฟังก์ชันที่มีโครงสร้างชัดเจน (`add`, `subtract`, `multiply`, `divide`, `power`) พร้อมการระบุประเภทข้อมูล (Type Hints), การใช้พารามิเตอร์เริ่มต้น (Default Parameters) และการเขียนคำอธิบายฟังก์ชันตามมาตรฐาน Docstring ของ PEP 257
2. **การควบคุมลำดับการทำงานแบบโต้ตอบ (Interactive Control Flow):** การเชื่อมโยงฟังก์ชันเข้ากับเมนูโต้ตอบผ่านลูป `while` พร้อมการเขียนโปรแกรมเชิงป้องกัน (Defensive Programming) เพื่อรับมือกับกรณีการหารด้วยศูนย์และข้อผิดพลาดจากการป้อนรูปแบบข้อมูลที่ไม่ถูกต้อง
3. **การเขียนโปรแกรมแบบแยกส่วน (Modular Programming):** การสร้างโมดูลอิสระ `my_utils.py` และการนำเข้า (import) มาใช้งานในแอปพลิเคชันหลักได้อย่างเป็นระเบียบและถูกต้อง
4. **การประยุกต์ใช้ไลบรารีมาตรฐานของ Python (Python Standard Library):** การนำโมดูลในตัวที่มีประสิทธิภาพสูงมาใช้ (`math`, `random`, `datetime`, `collections`, `pathlib`) เพื่อหลีกเลี่ยงการเสียเวลาเขียนโค้ดขึ้นมาใหม่โดยไม่จำเป็น (Don't reinvent the wheel)
5. **การปรับปรุงโครงสร้างโค้ด (Code Refactoring):** การแปลงโค้ดเดิมแบบรวมศูนย์เป็นก้อนเดียว (Monolithic Script) ให้กลายเป็นส่วนประกอบย่อยที่แยกส่วน สะอาด และสามารถทดสอบรายฟังก์ชัน (Unit Test) ได้ง่าย โดยยึดตามหลักการความรับผิดชอบเพียงอย่างเดียว (Single Responsibility Principle - SRP) และหลักการไม่เขียนโค้ดซ้ำซ้อน (DRY - Don't Repeat Yourself)

---
*(English Reference)*:
In this laboratory, we have comprehensively mastered:
1. **Function Design:** Encapsulating mathematical and logical routines into cleanly defined functions (`add`, `subtract`, `multiply`, `divide`, `power`) complete with type hints, default parameters, and PEP 257 docstrings.
2. **Interactive Control Flow:** Connecting functions to an interactive `while` loop menu, with defensive programming against zero-division and input format errors.
3. **Modular Programming:** Building an independent module `my_utils.py` and cleanly importing it in consumer applications.
4. **Python Standard Library:** Leveraging built-in modules (`math`, `random`, `datetime`, `collections`, `pathlib`) rather than reinventing the wheel.
5. **Code Refactoring:** Transforming legacy monolithic scripts into clean, modular, unit-testable components adhering to the Single Responsibility Principle (SRP) and DRY (Don't Repeat Yourself).
